```python
import matplotlib.pyplot as plt
import json

with open('test.json', 'r') as json_file:
    # Use json.load() to read the file and convert it to a dictionary
    both_dict = json.load(json_file)
```

```python
fig, ax = plt.subplots()

ax.scatter(both_dict['Tropics'].values(), both_dict['Arctic'].values())
for i, txt in enumerate(both_dict['Tropics'].keys()):
    ax.annotate(txt, (list(both_dict['Tropics'].values())[i], list(both_dict['Arctic'].values())[i]))
line = [-0.5, 1.5]
ax.plot(line, line)
```
